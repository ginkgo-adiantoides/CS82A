The best feed to use to maximize chicken weight is sunflower because it yields the highest average weight. While its range is one of the largest (197 lb), which means the weight of an individual chicken is hard to precisely predict, the minimum weight for this feed is the highest of all (226 lb). This means that even if the individual weights vary a lot, even the lowest likely weight would still be desirable. Contrast this with horsebean, which possesses a much tighter range (119 lb), but its maximum weight is a mere 227 lb: Nearly equal to the minimum weight of sunflower-fed chicks. A risk-averse strategy that specifically seeks to minimize range alone, and thus chooses horsebean, would poorly serve the goal of consistently high chick weights. Choosing sunflower would actually be much more risk-averse if risk-averse is defined as a high minimum weight rather than as a tight range.
